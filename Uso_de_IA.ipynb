# Uso de Inteligencia Artificial en el Proyecto


## Herramientas de IA utilizadas y tareas específicas
* **Gemini / ChatGPT:** Se utilizó principalmente como asistente de soporte y ayuda de código. Las tareas específicas incluyeron:
  * Resolución de conflictos de control de versiones (Git/GitHub) al momento de hacer *merge* entre las ramas de los distintos integrantes.
  * Solución de errores específicos del sistema operativo macOS (archivos `.DS_Store`) que bloqueaban el repositorio.
  * Asesoría lógica para la discretización de variables (creación de categorías balanceadas usando cuartiles en Pandas) para el modelo multiclase.
  * Formulación de argumentos técnicos basados en la metodología CRISP-DM para justificar la estructura del código (uso de notebooks vs. scripts modulares).

## Reflexión sobre el aprendizaje
El uso de la IA fue fundamental para no quedarnos estancados en errores del proceso que nos prohibian el avance optimo, especialmente en la integración del trabajo en equipo con Git. Cabe recalcar que no era solo copiar y pegar, se le pedía a la IA nos  explicara *por qué* ocurrían los errores (como los choques de archivos ocultos `.DS_Store` o `.idea/workspace.xml`) y cómo utilizar comandos como `git merge --abort` o `git rm` para resolverlos de forma segura. Con un entendimiento nos permitió que en casos futuros donde se presentaron errores como esos poder trabajarlos teniendo una guía y conocimiento sobre el problema.

Además, en temas de análisis de datos ayudó a comprender por que es bueno trabajar con cuartiles debido a que tenemos la certeza de que hay un balanceo de los datos, lo cual es vital para que el modelo de clasificación aprenda patrones reales y no genere conflictos
## Modificaciones realizadas al código generado por IA
* **Comandos de terminal:** Los comandos sugeridos por la IA se adaptaron a los nombres exactos de nuestras ramas (`app`, `data`, `main`) y a la estructura de carpetas de nuestro proyecto.
* **Lógica de preprocesamiento:** El código sugerido para la variable `nivel_demanda` no se pegó de forma aislada, sino que se integró estratégicamente dentro de las funciones de limpieza que ya había creado manualmente en `data_prep.py`, asegurando que no interfiriera con las variables *dummy* ni con el escalado (`MinMaxScaler`).
* **Manejo de Git:** Se modificó manualmente el archivo `.gitignore` del proyecto basándose en las recomendaciones de la IA para prevenir futuros conflictos con los editores de código del resto del equipo ya que se usan de ambos sistemas operativos.
