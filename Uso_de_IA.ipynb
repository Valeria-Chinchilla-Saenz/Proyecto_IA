# Uso de Inteligencia Artificial en el Proyecto

## Herramientas de IA utilizadas y tareas específicas
- **Gemini / ChatGPT / Claude:** Se utilizó principalmente como asistente de soporte, arquitectura y ayuda de código. Las tareas específicas incluyeron:
  - Resolución de conflictos de control de versiones (Git/GitHub) al momento de hacer *merge* entre las ramas de los distintos integrantes.
  - Solución de errores específicos del sistema operativo macOS (archivos `.DS_Store`) que bloqueaban el repositorio.
  - Asesoría lógica para la discretización de variables (creación de categorías balanceadas usando cuartiles en Pandas) para el modelo multiclase.
  - Formulación de argumentos técnicos basados en la metodología CRISP-DM para justificar la estructura del código (uso de notebooks vs. scripts modulares).
  - **Planificación y división equitativa del proyecto:** Análisis para dividir el trabajo entre los 3 integrantes (A, B, C) respetando las dependencias del pipeline (EDA -> Preprocesamiento -> Modelos -> API -> Frontend) y los porcentajes de la rúbrica del CUC.
  - **Evaluación de arquitectura (POO vs. Funciones):** Análisis conceptual sobre el uso de código modular funcional en `src/` frente a Programación Orientada a Objetos (POO) para preprocesamiento y entrenamiento.
  - **Diseño e integración del Frontend (Streamlit):** Maquetación de la arquitectura multipágina (`Home.py`, `1_Prediccion.py`, `2_Analisis.py`, `3_Metricas.py`) y conexión HTTP con la API REST en FastAPI mediante la librería `requests`.
  - **Análisis por Lotes y Lógica de Flota:** Desarrollo de la funcionalidad de carga de archivos CSV en `2_Analisis.py` para proyectar la demanda por franja horaria y generar recomendaciones automáticas de redistribución de bicicletas.
  - **Debugging e inyección de rutas:** Corrección del error `ModuleNotFoundError: No module named 'src'` mediante la inyección dinámica de la raíz del proyecto en `sys.path`.
  - **Formateo y post-procesamiento:** Implementación de reglas para redondear predicciones de regresión a enteros no negativos (`max(0, int(round(...)))`), e integración tabular del `classification_report` real en DataFrames de Pandas.

## Reflexión sobre el aprendizaje
El uso de la IA fue fundamental para no quedarnos estancados en errores del proceso que nos prohibían el avance óptimo, especialmente en la integración del trabajo en equipo con Git. Cabe recalcar que no era solo copiar y pegar, se le pedía a la IA nos explicara *por qué* ocurrían los errores (como los choques de archivos ocultos `.DS_Store` o `.idea/workspace.xml`) y cómo utilizar comandos como `git merge --abort` o `git rm` para resolverlos de forma segura. Con un entendimiento nos permitió que en casos futuros donde se presentaron errores como esos poder trabajarlos teniendo una guía y conocimiento sobre el problema.

Además, en temas de análisis de datos ayudó a comprender por qué es bueno trabajar con cuartiles debido a que tenemos la certeza de que hay un balanceo de los datos, lo cual es vital para que el modelo de clasificación aprenda patrones reales y no genere conflictos.

De igual forma, en el desarrollo del frontend e integración del sistema, la IA actuó como un tutor técnico. Nos permitió comprender la arquitectura cliente-servidor entre Streamlit y FastAPI, la forma en que Python gestiona la jerarquía de módulos (`sys.path`) en aplicaciones multipágina, y la importancia de evaluar críticamente las métricas de las redes neuronales (`Accuracy`, `Recall` y `F1-Score` en horas pico) para seleccionar el modelo óptimo. Esto nos demostró que la IA es una herramienta excelente para acelerar la maquetación y la resolución de errores sintácticos, pero que la validación lógica y la adaptación a los datos reales siempre dependen del criterio del desarrollador.

## Modificaciones realizadas al código generado por IA
- **Comandos de terminal:** Los comandos sugeridos por la IA se adaptaron a los nombres exactos de nuestras ramas (`app`, `data`, `main`) y a la estructura de carpetas de nuestro proyecto.
- **Lógica de preprocesamiento:** El código sugerido para la variable `nivel_demanda` no se pegó de forma aislada, sino que se integró estratégicamente dentro de las funciones de limpieza que ya había creado manualmente en `data_prep.py`, asegurando que no interfiriera con las variables *dummy* ni con el escalado (`MinMaxScaler`).
- **Manejo de Git:** Se modificó manualmente el archivo `.gitignore` del proyecto basándose en las recomendaciones de la IA para prevenir futuros conflictos con los editores de código del resto del equipo ya que se usan de ambos sistemas operativos.
- **Sincronización con FastAPI:** Se ajustaron manualmente los nombres de las claves al consumir el JSON retornado por la API (`resultado.get('prediccion')` / `resultado.get('cnt')`) para alinearlos con las variables de los esquemas Pydantic del backend.
- **Post-procesamiento de Regresión:** Se agregó la restricción lógica `max(0, int(round(...)))` sobre las predicciones continuas para evitar mostrar fracciones o valores negativos de bicicletas en la interfaz web.
- **Inyección de Rutas en Streamlit:** Se añadieron las líneas de resolución de rutas con `sys.path` al inicio de las páginas secundarias de Streamlit para solucionar los errores de importación del paquete `src`.
- **Integración de Métricas Reales:** Se reemplazaron todas las métricas de prueba iniciales por los datos reales calculados sobre el conjunto de prueba (`test.csv`) en nuestros notebooks (Accuracy del 76%/77%, soporte por clase y gráficos de curvas de pérdida y matriz de confusión).
