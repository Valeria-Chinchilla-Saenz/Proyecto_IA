```python
import pandas as pd
import pickle
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
df=pd.read_csv('../data/processed/datos_post_eda.csv')
```

```python
df = df.drop(['Unnamed: 0'], axis=1)
```

```python

df['nivel_demanda'] = pd.qcut(df['cnt'], q=4, labels=[0, 1, 2, 3])
```

```python
columnas_categoricas = ['season', 'hr', 'weathersit', 'weekday']
```

```python
df_procesado = pd.get_dummies(df, columns=columnas_categoricas, dtype=int)
```

```python
train_df, test_df = train_test_split(df_procesado, test_size=0.2, random_state=42)
```

```python
columnas_numericas = ['temp', 'atemp', 'hum', 'windspeed']
scaler = MinMaxScaler()
```

```python
train_df[columnas_numericas] = scaler.fit_transform(train_df[columnas_numericas])
test_df[columnas_numericas] = scaler.transform(test_df[columnas_numericas])
```

```python
os.makedirs('../models', exist_ok=True)
with open('../models/scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)
```

```python
X = train_df.drop(['cnt', 'nivel_demanda'], axis=1)
with open('../models/columnas.pkl', 'wb') as f:
    pickle.dump(X.columns.tolist(), f)
```

```python
os.makedirs('../data/processed', exist_ok=True)
train_df.to_csv('../data/processed/train.csv', index=False)
test_df.to_csv('../data/processed/test.csv', index=False)

print("¡Preprocesamiento completado! Archivos guardados en data/processed/")
```
